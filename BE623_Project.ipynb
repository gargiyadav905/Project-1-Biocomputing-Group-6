Question- 2a

```python
from Bio import SeqIO
record = SeqIO.read("sequence.gb", "genbank")

# To find the MT-ND1 CDS region
for feature in record.features:
    if feature.type == "CDS" and feature.qualifiers.get("gene", [""])[0] == "ND1":
        cds_length = feature.extract(record.seq)

        print("The CDS length is:", len(cds_length))
        print("Remainder when divided by 3:", len(cds_length) % 3)

        # It is given that 3' A completes the TAA stop codon so, I am adding that to the cds_length
        cds_length = cds_length + "A"

        print("New corrected CDS length is:", len(cds_length))
        print("Remainder after correction is:", len(cds_length) % 3)

        # Checking the last stop codon
        print("Last codon:", cds_length[-3:])
```

Question-2b

```python
cds_rna = str(cds_length).replace("T", "U")
genetic_code = { "UUU":"F", "UUC":"F", "UUA":"L", "UUG":"L", "UCU":"S", "UCC":"S", "UCA":"S", "UCG":"S", "UAU":"Y", "UAC":"Y", "UAA":"*", "UAG":"*", "UGU":"C", "UGC":"C", "UGA":"W", "UGG":"W",  "CUU":"L", "CUC":"L", "CUA":"L", "CUG":"L",  "CCU":"P", "CCC":"P", "CCA":"P", "CCG":"P",  "CAU":"H", "CAC":"H", "CAA":"Q", "CAG":"Q",  "CGU":"R", "CGC":"R", "CGA":"R", "CGG":"R", "AUU":"I", "AUC":"I", "AUA":"M", "AUG":"M", "ACU":"T", "ACC":"T", "ACA":"T", "ACG":"T", "AAU":"N", "AAC":"N", "AAA":"K", "AAG":"K", "AGU":"S", "AGC":"S", "AGA":"*", "AGG":"*", "GUU":"V", "GUC":"V", "GUA":"V", "GUG":"V", "GCU":"A", "GCC":"A", "GCA":"A", "GCG":"A", "GAU":"D", "GAC":"D", "GAA":"E", "GAG":"E", "GGU":"G", "GGC":"G", "GGA":"G", "GGG":"G" }
protein = ""

for i in range(0, len(cds_length), 3):
    codon = cds_rna[i:i+3]
    protein += genetic_code[codon]

print(" The first 30 residues are:", protein[:30])
print("The total protein length is:", len(protein))
print("last aa:", protein[-1])
```

Question-2c

```python
deposited_protein = str(SeqIO.read("aminoacid.fasta", "fasta").seq)
translated_protein = protein
#Removing the last stop codon as my translated protein has it
if translated_protein.endswith("*"):
    translated_protein = translated_protein[:-1]

if translated_protein == deposited_protein:
    print("Both the protein sequences are identical.")
else:
    print("Both the protein sequences are NOT identical.")
print("Translated protein length:", len(translated_protein))
print("Deposited protein length:", len(deposited_protein))
```

```python
from Bio.Seq import Seq
# Translating the CDS using Biopython
biopython_protein = str(Seq(cds_length).translate(table=2))

if protein == biopython_protein:
    print("My translation and Biopython translation agree.")
else:
    print("The translations do NOT agree.")

print("My translation length:", len(protein))
print("Biopython translation length:", len(biopython_protein))
print (protein[-1])
print (biopython_protein[-1])
```
